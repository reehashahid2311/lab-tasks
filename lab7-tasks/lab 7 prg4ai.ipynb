```python

class student:
    def __init__(self, name, age, course):
        self.name = name
        self.age = age
        self.course = course

    def display(self):
        print(self.name)
        print(self.age)
        print(self.course)

s1 = student("ali", 23, "python")
s1.display()
```

task no 2

```python
class graduate(student):
    def __init__(self,name, age, course,  research_topic):
        super().__init__(name , age, course)
        self.research_topic=research_topic

    def display1(self) :
        super().display()
        print(self.research_topic)
```

```python
g1=graduate("arshia",19,"ai","code")
g1.display1()
```
